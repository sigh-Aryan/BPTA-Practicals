```python
import networkx as nx
import matplotlib.pyplot as plt
import time
import random
import os


def generate_subsets(vertices):
    subsets = []

    def make_subsets(index, current):
        if index == len(vertices):
            subsets.append(current[:])
            return

        current.append(vertices[index])
        make_subsets(index + 1, current)
        current.pop()
        make_subsets(index + 1, current)

    make_subsets(0, [])
    return subsets


def is_vertex_cover(edges, subset):
    for u, v in edges:
        if u not in subset and v not in subset:
            return False
    return True


def vertex_cover(vertices, edges):
    subsets = generate_subsets(vertices)
    best_cover = vertices
    for subset in subsets:
        if len(subset) < len(best_cover):
            if is_vertex_cover(edges, subset):
                best_cover = subset

    return best_cover


# def random_graph(n, m):
#     edges = []
#     while len(edges) < m:
#         u = random.randint(1, n)
#         v = random.randint(1, n)
#         if u != v and (u, v) not in edges and (v, u) not in edges:
#             edges.append((u, v))

#     return edges

# above random\-graph(n,m) fxn doesn't take disconnected graphs into considertaion, below one does.
 

def random_graph(n, m):
    if m < n - 1:
        raise ValueError(
            f"connected graph with {n} vertices needs at least {n - 1} edges."
        )

    max_edges = n * (n - 1) // 2
    if m > max_edges:
        raise ValueError(
            f"simple graph with {n} vertices can have at most {max_edges} edges."
        )

    edges = set()
    vertices = list(range(1, n + 1))
    random.shuffle(vertices)
    for i in range(1, n):
        u = vertices[i]
        v = random.choice(vertices[:i])
        edges.add((min(u, v), max(u, v)))

    while len(edges) < m:
        u = random.randint(1, n)
        v = random.randint(1, n)
        if u == v:
            continue

        edge = (min(u, v), max(u, v))
        if edge not in edges:
            edges.add(edge)

    return list(edges)


def draw_graph(n, m, vertices, edges, cover):
    graph = nx.Graph()
    graph.add_nodes_from(vertices)
    graph.add_edges_from(edges)
    colors = []
    for vertex in vertices:
        if vertex in cover:
            colors.append("red")
        else:
            colors.append("blue")

    plt.figure(figsize=(10, 10))
    pos = nx.spring_layout(graph, seed=42)

    nx.draw(
        graph,
        pos,
        with_labels=True,
        node_color=colors,
        node_size=1000,
        font_color="white"
    )

    plt.title(f"Minimum Vertex Cover\nn = {n}, m = {m}\nCover = {cover}")
    filename = f"all_outputs/images/graph_n{n}_m{m}.png"
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.close()

os.makedirs("all_inputs", exist_ok=True)
os.makedirs("all_outputs/images", exist_ok=True)


n = 10
m_values = [10, 15, 20, 25, 30, 35, 40, 45]

for i, m in enumerate(m_values, start=1):
    vertices = list(range(1, n + 1))
    edges = random_graph(n, m)
    input_filename = f"all_inputs/input{i}.txt"
    with open(input_filename, "w") as input_file:
        input_file.write(f"{n} {m}\n")
        for u, v in edges:
            input_file.write(f"{u} {v}\n")

for i in range(1, 9):
    input_filename = f"all_inputs/input{i}.txt"
    with open(input_filename, "r") as input_file:
        first_line = input_file.readline().split()
        n = int(first_line[0])
        m = int(first_line[1])
        vertices = list(range(1, n + 1))
        edges = []
        for line in input_file:
            u, v = map(int, line.split())
            edges.append((u, v))

    start = time.perf_counter()
    cover = vertex_cover(vertices, edges)
    end = time.perf_counter()
    time_taken = end - start
    output_filename = f"all_outputs/output{i}.txt"
    with open(output_filename, "w") as output:
        output.write(f"Number of Nodes = {n}\n")
        output.write(f"Number of Edges = {m}\n\n")
        output.write("Edge List:\n")
        for u, v in edges:
            output.write(f"{u} {v}\n")

        output.write("\n")
        output.write(f"Vertex Cover Size = {len(cover)}\n")
        output.write(f"Vertex Cover Nodes = {cover}\n")
        output.write(f"Running Time = {time_taken:.10f} seconds\n")


    draw_graph(n, m, vertices, edges, cover)


    print(f"Input File = input{i}.txt")
    print(f"n = {n}, m = {m}")
    print(f"Vertex Cover = {cover}")
    print(f"Size = {len(cover)}")
    print(f"Running Time = {time_taken:.10f} seconds")
    print("-" * 40)
```
