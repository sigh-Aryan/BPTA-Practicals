```python
import networkx as nx
import matplotlib.pyplot as plt

class Graph:

    def __init__(self, vertices):
        self.V = vertices
        self.edges = set()

    def add_edge(self, u, v):
        edge = (min(u, v), max(u, v))

        if edge not in self.edges:
            self.edges.add(edge)

    def generate_subsets(self, index, current):
        if index == self.V:
            print(current)
            return

        current.append(index)
        self.generate_subsets(index + 1, current)

        current.pop()
        self.generate_subsets(index + 1, current)

n = int(input("Enter number of vertices (n): "))
m = int(input("Enter number of edges (m): "))

graph = Graph(n)

print("Enter the edges (u v):")

for i in range(m):
    print(f"Edge {i + 1}:")
    u, v = map(int, input().split())

    graph.add_edge(u, v)


print("\ngraph edges:")
for edge in sorted(graph.edges):
    print(edge)


print("\nall possible subsets:")
graph.generate_subsets(0, [])

G = nx.Graph()

G.add_nodes_from(range(graph.V))
G.add_edges_from(graph.edges)

nx.draw(
    G,
    with_labels=True,
    node_size=1500,
    font_size=12
)

plt.title(f"(n={n}, m={m})")
plt.show()
```
