```python
import random
import networkx as nx
import matplotlib.pyplot as plt

class Graph:
    def __init__(self, vertices):
        self.V= vertices
        self.edges= set()

    def add_edge(self, u, v):
        edge= (min(u, v), max(u, v))

        if edge not in self.edges:
            self.edges.add(edge)

    def generate_subsets(self, index, current):
        if index== self.V:
            print(current)
            return

        current.append(index)
        self.generate_subsets(index+1, current)

        current.pop()
        self.generate_subsets(index+1, current)


def generate_random_graph(n, m):
    graph= Graph(n)

    while len(graph.edges)< m:
        u=random.randint(0, n- 1)
        v= random.randint(0, n- 1)

        if u!=v:
            graph.add_edge(u, v)

    return graph


n= 10
m_values= [10, 15, 20, 25, 30, 35, 40, 45]
m= random.choice(m_values)

graph= generate_random_graph(n, m)

print("graph edges:")
for edge in sorted(graph.edges):
    print(edge)

print("\nall possible subsets:")
graph.generate_subsets(0, [])

G= nx.Graph()
G.add_nodes_from(range(graph.V))
G.add_edges_from(graph.edges)

nx.draw(G, with_labels=True, node_size=1500, font_size=12)
plt.title(f"(n={n}, m={m})")
plt.show()
```
